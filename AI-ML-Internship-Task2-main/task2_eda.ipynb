## Exploratory Data Analysis (EDA)

Explore the Titanic dataset using statistics, visualizations, correlation analysis, and skewness.

## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

## Step 2: Load Dataset

```python
# Replace with your dataset path
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

# Display first few rows
df.head()
```

## Step 3: Summary Statistics

```python
# Basic info
df.info()

# Descriptive statistics
df.describe()

# Check for missing values
df.isnull().sum()
```

What you'll learn: mean, median, standard deviation, min, max, percentiles, missing values.

## Step 4: Univariate Analysis

### 1. Histograms

```python
# Numeric features histogram
df.hist(bins=20, figsize=(12,10))
plt.show()
```

### 2. Boxplots

```python
# Boxplot for Age
sns.boxplot(x='Age', data=df)
plt.show()

# Boxplot for Fare by Pclass
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.show()
```

Why boxplots: Identify outliers, understand data spread, compare distributions across categories.

## Step 5: Bivariate/Multivariate Analysis

### 1. Pairplot

```python
sns.pairplot(df[['Age','Fare','Pclass','Survived']], hue='Survived')
plt.show()
```

### 2. Correlation Matrix

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.show()
```

Why correlation: Measures linear relationship between numeric features. Helps in feature selection and multicollinearity detection.

## Step 6: Detect Skewness

```python
# Check skewness
df[['Age','Fare']].skew()

# Optional: visualize skewness
sns.histplot(df['Fare'], kde=True)
plt.show()
```

Skewness: Positive skew → tail on right, Negative skew → tail on left.

## Step 7: Interactive Visualization (Plotly)

```python
fig = px.box(df, x='Pclass', y='Fare', color='Survived', title='Fare Distribution by Class')
fig.show()
```

## Step 8: Basic Observations / Feature-Level Inferences

- Older passengers tend to pay higher fare.

- Most passengers are in Pclass 3 (lower class).

- There are some outliers in Age and Fare.

- Positive correlation between Fare and Pclass (as expected).
